```python
import pandas as pd

url = 'https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/mtcars.csv'
df = pd.read_csv(url)
```

```python
print(df.head())
print(df.describe())
print(df.info())
```

```python
df = df.rename(columns={'Unnamed: 0': 'car_model'})
```

```python
df_subset = df[['mpg', 'hp', 'wt', 'cyl']].copy()
```

```python
df_subset = df_subset.rename(columns={'cyl': 'cylinders'})
```

```python
print(df_subset.describe())
average_mpg = df_subset['mpg'].mean()
print(f"Average MPG: {average_mpg}")
```

```python
four_cyl_vehicles = df_subset[df_subset['cylinders'] == 4]
print(four_cyl_vehicles.describe())
```
